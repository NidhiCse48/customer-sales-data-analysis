```python

import pandas as pd

sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone",
                "Tablet", "Laptop", "Phone", "Tablet", "Laptop"],
    "Category": ["Electronics"] * 10,
    "Quantity": [2, 5, 3, 1, 4, 2, 3, 6, 1, 2],
    "Unit_Price": [50000, 20000, 15000, 50000, 20000,
                   15000, 50000, 20000, 15000, 50000],
    "Region": ["North", "South", "East", "West", "North",
               "South", "East", "West", "North", "South"]
})

sales["Revenue"] = sales["Quantity"] * sales["Unit_Price"]

display(sales)
```

```python

product_sales = sales.groupby("Product")["Revenue"].sum()
print(product_sales)
```

```python

product_sales = sales.groupby("Product")["Revenue"].sum()
print(product_sales)
```

```python

import matplotlib.pyplot as plt

product_sales.sort_values(ascending=False).plot(kind="bar")
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

region_sales = sales.groupby("Region")["Revenue"].sum()

print("Region-wise Revenue:")
print(region_sales.sort_values(ascending=False))
```

```python

import matplotlib.pyplot as plt

region_sales.sort_values(ascending=False).plot(kind="bar")

plt.title("Region-wise Sales Revenue")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

best_region = region_sales.idxmax()
highest_revenue = region_sales.max()

print("Best Performing Region:", best_region)
print("Highest Revenue:", highest_revenue)
```

```python

sales["Order_Date"] = pd.to_datetime([
    "2026-01-05", "2026-01-12", "2026-02-03",
    "2026-02-15", "2026-03-02", "2026-03-18",
    "2026-04-04", "2026-04-21", "2026-05-10",
    "2026-05-25"
])

sales["Month"] = sales["Order_Date"].dt.strftime("%b")

monthly_sales = sales.groupby("Month")["Revenue"].sum()

print("Monthly Revenue:")
print(monthly_sales)
```

```python

cost_prices = {
    "Laptop": 35000,
    "Phone": 12000,
    "Tablet": 9000
}

sales["Unit_Cost"] = sales["Product"].map(cost_prices)

sales["Total_Cost"] = sales["Quantity"] * sales["Unit_Cost"]

sales["Profit"] = sales["Revenue"] - sales["Total_Cost"]

display(sales[[
    "Product", "Quantity", "Revenue",
    "Total_Cost", "Profit"
]])
```

```python

product_profit = sales.groupby("Product")["Profit"].sum()

print("Product-wise Profit:")
print(product_profit.sort_values(ascending=False))
```

```python

import matplotlib.pyplot as plt

product_profit.sort_values().plot(
    kind="barh",
    figsize=(7, 4)
)

plt.title("Total Profit by Product")
plt.xlabel("Profit (₹)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python

print("CUSTOMER SALES ANALYSIS REPORT")
print("-" * 35)
print("Total Revenue: ₹", sales["Revenue"].sum())
print("Total Cost: ₹", sales["Total_Cost"].sum())
print("Total Profit: ₹", sales["Profit"].sum())
print("Total Units Sold:", sales["Quantity"].sum())

print(
    "Most Profitable Product:",
    product_profit.idxmax()
)
```

```python

sales.to_csv("customer_sales_analysis.csv", index=False)

print("Final dataset saved successfully!")
```

```python

sales["Profit_Margin_Percent"] = (
    sales["Profit"] / sales["Revenue"]
) * 100

display(sales[[
    "Product", "Revenue", "Profit", "Profit_Margin_Percent"
]])
```

```python

total_revenue = sales["Revenue"].sum()
total_profit = sales["Profit"].sum()
total_units = sales["Quantity"].sum()

overall_margin = (total_profit / total_revenue) * 100

print("===== SALES DASHBOARD KPIs =====")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Units Sold: {total_units}")
print(f"Overall Profit Margin: {overall_margin:.2f}%")
```
